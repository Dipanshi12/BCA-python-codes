```python
n = int(input("Enter a number:"))

square = n * n

temp = n
count = 0

while temp > 0:
    temp = temp // 10
    count += 1

if square % (10 ** count) == n:
    print("Automorphic Number")
else:
    print("Not an Automorphic Number")
```

```python
n = int(input("enter a number:"))
temp = n
strong_sum = 0
while temp>0:
    digit = temp % 10
    fact = 1
    for i in range(1, digit+1):
      fact = fact * i
        
    strong_sum = strong_sum + fact
    temp = temp // 10 
if strong_sum == n:
    print("strong number")
else:
    print("not strong number:")
```
