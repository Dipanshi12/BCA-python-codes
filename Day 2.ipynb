```python
num = int(input("enter a number:"))
temp = num
digits = len(str(num))
sum = 0

while temp>0:
    digit = temp % 10
    sum = sum + (digit ** digits)
    temp = temp // 10
   
if sum == num:
    print("armstrong")
else:
    print("not armstrong")
    
```

```python
num = int(input("enter a number:"))
temp = num
reverse = 0
while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp = temp // 10
print("reverse:", reverse)
```
