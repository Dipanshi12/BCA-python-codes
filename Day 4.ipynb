```python
weight = float(input("enter weight of the body (kg):"))
height = float(input("enter height of the body (m):"))
bmi = weight/(height * height)
print("BMI", bmi)
if bmi < 18.5:
    print("underweight")
elif bmi < 25:
    print("normal")
elif bmi < 30:
    print("overweight")
else:
    print("obses")
```
